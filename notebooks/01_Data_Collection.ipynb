```python
url = "https://api.fda.gov/drug/label.json?limit=50"

response = requests.get(url)
data = response.json()
```

```python
drug_data = []

for drug in data["results"]:
    openfda = drug.get("openfda", {})

    record = {
        "brand_name": openfda.get("brand_name", ["N/A"])[0],
        "generic_name": openfda.get("generic_name", ["N/A"])[0],
        "manufacturer": openfda.get("manufacturer_name", ["N/A"])[0],
        "purpose": drug.get("purpose", ["N/A"])[0]
    }

    drug_data.append(record)
```
